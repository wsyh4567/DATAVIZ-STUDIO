# D-Tale 数据探索示例

本笔记本展示了如何使用 `D-Tale` 库在 Jupyter 或 VS Code 中进行强大的交互式数据分析。

```python
# 1. 安装 d-tale (如果尚未安装)
# !pip install dtale
```

```python
import pandas as pd
import dtale
import seaborn as sns

# 加载示例数据集 (使用 Seaborn 的泰坦尼克数据集)
df = sns.load_dataset('titanic')

print("数据集已加载，形状为:", df.shape)
df.head()
```

```python
# 2. 在笔记本内部启动 D-Tale
d = dtale.show(df)

# 如果你在 VS Code 或 Jupyter 中运行，它会直接显示交互式界面
d
```
